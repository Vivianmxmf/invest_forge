```python
import sys, pathlib
ROOT = pathlib.Path.cwd().parent
if str(ROOT) not in sys.path: sys.path.insert(0, str(ROOT))
from janus_terminal.tools.data_tools import FakeDataProvider
fake = FakeDataProvider(sample_dir=ROOT/'data'/'sample')
for code in ['688981.SH', '600519.SH', '300750.SZ', '600276.SH', '601398.SH']:
    fm = fake.fundamentals(code)
    print(f'{code}  ROE={fm.roe:.1%}  net_margin={fm.net_margin:.1%}  PE={fm.pe_ratio}')
```

```python
# Simple DuPont 3-factor decomposition (illustrative)
fm = fake.fundamentals('688981.SH')
# leverage proxy from debt_ratio
leverage = 1.0 / (1.0 - fm.debt_ratio)
asset_turnover = fm.net_margin and (fm.roe / fm.net_margin / leverage) or 0.0
print(f'ROE={fm.roe:.2%}  =  net_margin({fm.net_margin:.2%}) × turnover({asset_turnover:.2f}) × leverage({leverage:.2f})')
```
