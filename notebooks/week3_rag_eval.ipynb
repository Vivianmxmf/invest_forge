# Week 3 — Hybrid RAG smoke + stub RAGAS

Exercises the in-memory hybrid retriever on the synthetic KB and runs the offline ``stub_evaluate``.  No Qdrant / no OpenAI key needed.

```python
import sys, pathlib
ROOT = pathlib.Path.cwd().parent
if str(ROOT) not in sys.path: sys.path.insert(0, str(ROOT))
from invest_forge.knowledge_base.builder import build_in_memory
kb = build_in_memory(ROOT/'data'/'sample'/'kb')
print('chunks:', len(kb.texts))
```

```python
for q in ['净利率', '客户集中度', '锂价波动', '净息差']:
    hits = kb.retrieve(q, k=2)
    print(q, '→', [(h.source, round(h.score, 2)) for h in hits])
```

```python
from invest_forge.eval.ragas_eval import stub_evaluate
answers = ['公司 2024 年净利率 15.1%, 营收 432 亿元.']
ctxs = [[kb.retrieve('净利率', k=1)[0].text]]
print(stub_evaluate(answers, ctxs))
```
